Heatmat plot

```python

import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

numeric_data = tips.select_dtypes(include="number")

correlation = numeric_data.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Tips Dataset Correlation Heatmap")

plt.show()
```

practices question
Create a heatmap using the tips dataset.

Create a heatmap with annot=True and cmap="viridis".

Create a heatmap using the Iris dataset's numerical columns.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Datasets load 
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")

# 1. Create a heatmap using the tips dataset
plt.figure(figsize=(6, 4))
# Numeric columns ki correlation calculate 
tips_corr = tips.select_dtypes(include=["float64", "int64"]).corr()
sns.heatmap(data=tips_corr)
plt.title("1. Tips Dataset Heatmap")
plt.show()

# 2. Create a heatmap with annot=True and cmap="viridis"
plt.figure(figsize=(6, 4))
sns.heatmap(data=tips_corr, annot=True, cmap="viridis")
plt.title("2. Tips Heatmap with Annotations & Viridis Palette")
plt.show()

# 3. Create a heatmap using the Iris dataset's numerical columns
plt.figure(figsize=(6, 4))
iris_corr = iris.drop(columns=["species"]).corr()
sns.heatmap(data=iris_corr, annot=True, cmap="coolwarm")
plt.title("3. Iris Numerical Columns Heatmap")
plt.show()
```

Mini Project – Student Performance Visualization

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create student dataset
data = {
    "Student": ["Ali", "Ahmed", "Hasan", "Sara", "Ayesha", "Usman"],
    "Math": [85, 70, 92, 88, 95, 65],
    "Science": [80, 75, 90, 85, 92, 60],
    "English": [78, 82, 88, 90, 95, 70]
}

df = pd.DataFrame(data)

# Calculate average marks
df["Average"] = df[["Math", "Science", "English"]].mean(axis=1)

# 1. Barplot - Student Average Marks
sns.barplot(data=df, x="Student", y="Average")
plt.title("Student Average Marks")
plt.show()

# 2. Scatterplot - Math vs Science
sns.scatterplot(data=df, x="Math", y="Science")
plt.title("Math vs Science Marks")
plt.show()

# 3. Histogram - Average Marks Distribution
sns.histplot(data=df, x="Average", bins=5)
plt.title("Average Marks Distribution")
plt.show()

# 4. Boxplot - Average Marks
sns.boxplot(data=df, y="Average")
plt.title("Student Average Marks Distribution")
plt.show()

# 5. Heatmap - Marks Correlation
correlation = df[["Math", "Science", "English"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Marks Correlation Heatmap")
plt.show()

# Display student results
print(df)
```
