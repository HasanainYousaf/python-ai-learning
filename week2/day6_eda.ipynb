Handling missing value
1 remove missing value
2 fill missing values
3 fill with columns mean

```python
data = {
    "Name": ["Ali", "Ahmed", "Hasan", "Sara", "Ayesha"],
    "Math": [85, None, 92, 88, 95],
    "Science": [80, 75, None, 85, 92],
    "English": [78, 82, 88, None, 95]
}
df=pd.DataFrame(data)
print(df)
print(df.isnull())
print(df.isnull().sum())
print(df.isnull().sum().sum())
# Remove rows containing missing values
df_clean = df.dropna()

print(df_clean)
# Fill missing values with zero
df_filled = df.fillna(0)

print(df_filled)
# Fill missing Math marks with mean
df["Math"] = df["Math"].fillna(df["Math"].mean())

print(df)
```

Data Filtering and Outlier Detection

```python

import pandas as pd

# Create dataset
data = {
    "Name": ["Ali", "Ahmed", "Hasan", "Sara", "Ayesha"],
    "Math": [85, 70, 92, 88, 95],
    "Science": [80, 75, 90, 85, 92]
}

# Create DataFrame
df = pd.DataFrame(data)

# Students with Math marks greater than 80
high_marks = df[df["Math"] > 80]

print(high_marks)

# Students with Math marks less than 80
low_marks = df[df["Math"] < 80]

print(low_marks)

# Students with Math marks greater than 80
# AND Science marks greater than 85
result = df[(df["Math"] > 80) & (df["Science"] > 85)]

print(result)
```

CORELATION 

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create student dataset
data = {
    "Math": [50, 60, 70, 80, 90],
    "Science": [55, 65, 75, 85, 95],
    "English": [90, 80, 70, 60, 50]
}

# Create DataFrame
df = pd.DataFrame(data)

# Calculate correlation
correlation = df.corr()

# Display correlation
print(correlation)

# Create heatmap
sns.heatmap(correlation, annot=True, cmap="coolwarm")

# Add title
plt.title("Student Marks Correlation")

# Display graph
plt.show()
```

Feature Scaling
1 Normalization(min-maxscaling)
2 standardization(z-score scaling)

```python

import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create dataset
data = {
    "Age": [18, 20, 22, 24, 26],
    "Marks": [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

# Create scaler
scaler = MinMaxScaler()

# Apply normalization
scaled_data = scaler.fit_transform(df)

# Convert into DataFrame
df_scaled = pd.DataFrame(scaled_data, columns=df.columns)

# Display results
print(df_scaled)
```

```python

import pandas as pd
from sklearn.preprocessing import StandardScaler

# Create dataset
data = {
    "Age": [18, 20, 22, 24, 26],
    "Marks": [40, 50, 60, 70, 80]
}

df = pd.DataFrame(data)

# Create scaler
scaler = StandardScaler()

# Apply standardization
scaled_data = scaler.fit_transform(df)

# Convert into DataFrame
df_scaled = pd.DataFrame(scaled_data, columns=df.columns)

# Display results
print(df_scaled)
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 1. Create dataset
df = pd.DataFrame({
    "Age":    [22, 25, 30, 35, 40, 45, 50, 55],
    "Salary": [25000, 32000, 45000, 52000, 60000, 75000, 90000, 120000]
})
print(df)

# 2. Normalization (MinMaxScaler)
mm = MinMaxScaler()
df_norm = pd.DataFrame(mm.fit_transform(df), columns=["Age_norm", "Salary_norm"])

# 3. Standardization (StandardScaler)
ss = StandardScaler()
df_std = pd.DataFrame(ss.fit_transform(df), columns=["Age_std", "Salary_std"])

# 4. Original vs Normalized
print(pd.concat([df, df_norm.round(3)], axis=1))

# 5. Original vs Standardized
print(pd.concat([df, df_std.round(3)], axis=1))
```

Mini Project – Student Performance Data Analysis
. Project Objective
Analyze student performance and perform:
- Dataset understanding
- Missing values analysis
- Duplicate detection
- Data filtering
- Outlier visualization
- Correlation analysis
- Feature scaling

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# 1. Create student dataset
data = {
    "Name": ["Ali", "Ahmed", "Hasan", "Sara", "Ayesha", "Usman", "Zain"],
    "Math": [85, 70, 92, 88, 95, 65, 90],
    "Science": [80, 75, 90, 85, 92, 60, 88],
    "English": [78, 82, 88, 90, 95, 70, 85]
}

df = pd.DataFrame(data)

# 2. Calculate average marks
df["Average"] = df[["Math", "Science", "English"]].mean(axis=1)

# 3. Understand dataset
print("\nDataset:")
print(df)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# 6. Filter high-performing students
high_performers = df[df["Average"] >= 85]

print("\nHigh Performing Students:")
print(high_performers)

# 7. Visualize average marks
sns.barplot(data=df, x="Name", y="Average")
plt.title("Student Average Marks")
plt.show()

# 8. Detect outliers using boxplot
sns.boxplot(data=df, y="Math")
plt.title("Math Marks Outlier Analysis")
plt.show()

# 9. Correlation analysis
correlation = df[["Math", "Science", "English"]].corr()

print("\nCorrelation:")
print(correlation)

sns.heatmap(correlation, annot=True, cmap="coolwarm")
plt.title("Subject Correlation")
plt.show()

# 10. Feature scaling
scaler = MinMaxScaler()

df_scaled = pd.DataFrame(
    scaler.fit_transform(df[["Math", "Science", "English"]]),
    columns=["Math", "Science", "English"]
)

print("\nNormalized Marks:")
print(df_scaled)

# 11. Final results
print("\nFinal Student Results:")
print(df)
```
